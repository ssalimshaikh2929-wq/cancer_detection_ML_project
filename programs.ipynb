```python
#q1: Find the Highest Value from below List 

l = [20, 40, 10, 50, 30]

#solution-1
print(max(l))

#solution-2

highest = l[0]

for i in l:
    if i > highest:
        highest=i
print(highest)

#solution-3
l.sort()
print(l[-1])

```

```python


l = [20, 40, 10, 50, 30]

#solution-1
print(min(l))

#solution-2
l.sort()
print(l[0])

#solution-3

smallest=l[0]

for i in l:
    if i<smallest:
        smallest=i
print(smallest)
```
