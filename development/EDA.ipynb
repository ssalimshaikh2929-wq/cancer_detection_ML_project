```python
d = {'A': 50,
     'B': 100,
     'C' : 5}

lowest = max(d.values())
lowest_key = ''

for key in d:
    if d[key] < lowest:       #Here d[key] means value of each key
                              #Loop 1: key = A, d[key] i.e d['A'] = 50, 50<lowest, 50<100, True, lowest = 50
                              #Loop 2: key = B, d[key] i.e d['B'] = 100, 100<lowest, 100<50, False, lowest = 50
                              #Loop 3: key = C, d[key] i.e d['B'] = 100, 100<lowest, 100<100, True, lowest = 5
        lowest = d[key]      
        lowest_key = key

print(lowest)
print(lowest_key)
```

```python
L1 = [10, 20, 30, 40]
L2 = [30, 40, 50, 60]

#Solution:

Result_L1 = [ ]
Result_L2 = [ ]

for i in L1:
    if i not in L2:
        Result_L1.append(i)

for i in L2:
    if i not in L1:
        Result_L2.append(i)

print(Result_L1)
print(Result_L2)
print(Result_L1 + Result_L2)
```

```python
l = [1,2,4,8,64]

n = len(l)
diff = l[1]/l[0]

for i in range(1,n):
    if l[i+1] / l[i] != diff:
        l.insert(i+1, l[i]*diff)

print(l)
```
